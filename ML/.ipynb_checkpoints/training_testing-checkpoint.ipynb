```python
import pandas as pd
```

```python
df=pd.read_csv("final_data.csv")
df.head()
```

```python
df['Attack_Type'].unique()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
x.head()
```

```python
X=x.values
```

```python
y.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
le.fit(y)
y=le.transform(y)
```

```python
from collections import Counter
```

```python
Counter(y)
```

```python
from imblearn.over_sampling import RandomOverSampler
```

```python
ros = RandomOverSampler(random_state=42)
X_res, y_res = ros.fit_resample(X, y)
```

```python
Counter(y_res)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train_val, X_test, y_train_val, y_test = train_test_split(X_res, y_res, test_size=0.2, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_train_val, y_train_val, test_size=0.2, random_state=42)
```

```python
from keras.models import Sequential
from keras.layers import Conv1D, MaxPooling1D, Flatten, Dense

# define the model architecture
model = Sequential()
model.add(Conv1D(32, kernel_size=3, activation='relu', input_shape=(X_train.shape[1],1)))
model.add(MaxPooling1D(pool_size=2))
model.add(Flatten())
model.add(Dense(10, activation='relu'))
model.add(Dense(5, activation='softmax'))

# compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X_train.reshape((X_train.shape[0], X_train.shape[1], 1)), y_train, 
          validation_data=(X_val.reshape((X_val.shape[0], X_val.shape[1], 1)), y_val), 
          epochs=10, batch_size=32)
```

```python
model.save("model.h5")
```

```python
import joblib
```

```python
joblib.dump(le,"label.joblib")
```

```python
test_dat=df[df['Attack_Type']=='r2l'].iloc[10:11,:-1]
test_dat
```

```python
test_dat.to_csv("r2l.csv",index=False)
```

```python
model.evaluate(X_test.reshape((X_test.shape[0], X_test.shape[1], 1)), y_test)
```
